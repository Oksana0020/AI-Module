```python
from sklearn.datasets import load_iris
```

```python
iris_data = load_iris()
```

```python
iris_data['data']
```

```python
iris_data['target']
```

```python
iris_data['target_names']
```

```python
from keras.models import Sequential
from keras.layers import Dense
```

```python
model = Sequential()
model.add(Dense(10, input_shape=(4,), activation='relu'))
model.add(Dense(5, activation='relu'))
model.add(Dense(3, activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(optimizer='adam', loss='categorical_crossentropy', metrics=['accuracy'])
```

```python
X = iris_data['data']
y = iris_data['target']
```

```python
y.shape
```

```python
X.shape
```

```python
from tensorflow.keras.utils import to_categorical
```

```python
Y = to_categorical(y)
```

```python
fit_data = model.fit(X, Y,epochs=500)
```

```python
import matplotlib.pyplot as plt
plt.plot(fit_data.history['loss'])
plt.plot(fit_data.history['accuracy'])
```

```python
from sklearn.model_selection import train_test_split
```

```python
help(train_test_split)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, Y, test_size=0.2)
```

```python
y_train
```

```python
hist = model.fit(X_train, y_train, epochs=200)
```

```python
plt.plot(hist.history['loss'])
plt.plot(hist.history['accuracy'])
```

```python
y_pred = model.predict(X_test)
```

```python
y_test
```

```python
y_pred
```

```python
from sklearn.metrics import confusion_matrix
```

```python
help(confusion_matrix)
```

```python
confusion_matrix(y_test.argmax(axis=1), y_pred.argmax(axis=1))
```

```python
hist2 = model.fit(X_train, y_train,  validation_data=(X_test,y_test), epochs=200)
```

```python
plt.plot(hist2.history['loss'])
plt.plot(hist2.history['accuracy'])
plt.plot(hist2.history['val_loss'])
plt.plot(hist2.history['val_accuracy'])
```
